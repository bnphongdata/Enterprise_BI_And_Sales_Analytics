```python
# Bài 1: Khởi tạo biến cơ bản

# 1. Tạo 3 biến a, b, c
# 2. Gán giá trị số nguyên cho các biến
# 3. In giá trị biến b ra màn hình

a = 7
b = 3
c = 4

print(b)

# Bài 2: Kiểm tra kiểu dữ liệu
# Dùng hàm type() để xem chuỗi thuộc kiểu dữ liệu gì

type("-7")

# Bài 3: Kiểu dữ liệu Đúng/Sai (Boolean)
# Kiểm tra kiểu dữ liệu của giá trị True/False

type(True)

# Bài 4: Ép kiểu dữ liệu (Type Casting)
# Chuyển đổi một chuỗi số sang số nguyên (int)

int("-7")

# Bài 5: Nối chuỗi cơ bản
# Tạo 2 biến chuỗi rồi nối chúng lại với nhau

a = "Net"
b = "Profit"

print(a + " " + b)

# Bài 6: Nối chuỗi với số
# Ghép một số với một chuỗi ký tự
# Nhớ ép số về chuỗi bằng str() trước khi nối để không bị lỗi

a = 5
b = "Profit"

print(str(a) + " " + b)

# Bài 7: Các phép toán số học cơ bản
# Thực hiện chia (/), cộng (+), và trừ (-) hai biến số

a = 1
b = 2

print(a / b)
print(a + b)
print(a - b)

# Bài 8: Toán tử logic HOẶC (or)
# Chỉ cần 1 trong các điều kiện đúng là kết quả trả về True

x = 0
x < 1 or x < 3

# Bài 9: Toán tử logic VÀ (and)
# Cả hai điều kiện đều phải đúng thì mới trả về True

a = 19
b = 17
c = 1

a + b > c - a and b + c < a

# Bài 10: Các hàm tích hợp sẵn min() và max()
# Tìm số lớn nhất và nhỏ nhất trong tập hợp số

print(max(1, 2, 3, 5))
print(min(1, 2, 3, 5))

# Bài 11: Lấy trị tuyệt đối và làm tròn số
# Dùng hàm abs() lấy trị tuyệt đối và round() để làm tròn thập phân

abs(-3)
round(4.5565755, 2)

# Bài 12: Tính tổng và đếm số phần tử
# Dùng sum() tính tổng và len() đếm số lượng phần tử trong danh sách

sum([1, 2, 3, 4, 5])
```
