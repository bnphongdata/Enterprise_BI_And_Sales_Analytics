```python
# Đảo ngược thứ tự các phần tử trong danh sách (list)

number_list = [1, 2, 3, 4, 6]
number_list.reverse()
print(number_list)
```

```python
# Sắp xếp danh sách giảm dần (nếu bỏ reverse=True thì mặc định xếp tăng dần)

number_list = [5, 2, 8, 1, 9, 6]
number_list.sort(reverse=True)
print(number_list)
```

```python
# Thêm một phần tử vào cuối danh sách bằng append()

number_list = [5, 2, 8, 1, 9, 6]
number_list.append(10)
print(number_list)
```

```python
# Nối thêm nhiều phần tử vào danh sách bằng extend()

number_list = [5, 2, 8, 1, 9, 6]
number_list.extend([1, 2, 3, 5])
print(number_list)


```

```python
# Chèn phần tử vào một vị trí cụ thể bằng insert()

number_list = [5, 2, 8, 1, 9, 6]
number_list.insert(0, 11)
print(number_list)

```

```python
# Gán đè giá trị mới cho phần tử theo vị trí chỉ mục (index)

number_list = [5, 2, 8, 1, 9, 6]
number_list[0] = "number"
print(number_list)
```

```python
# Xóa phần tử theo vị trí chỉ mục bằng pop() hoặc del

number_list = [5, 2, 8, 1, 9, 6]
del number_list[1]
print(number_list)
```

```python
# Xóa phần tử theo giá trị bằng remove()

number_list = [5, 2, 8, 1, 9, 6]
number_list.remove(2)
print(number_list)
```

```python
# Lấy phần tử ra khỏi danh sách và lưu giá trị đó lại bằng pop()

number_list = [5, 2, 8, 1, 9, 6]
x = number_list.pop(0)
print(number_list)
print("Removed value:", x)

```

```python
# Cắt lát danh sách (Slicing): trích xuất một đoạn danh sách con

number_list = [5, 2, 8, 1, 9, 6, 4, 5, 9, 11, 12, 13]

# Lấy từ đầu danh sách đến index 4 (gồm các phần tử từ 0 đến 4)
print(number_list[:5])

# Lấy từ index 3 đến hết danh sách
print(number_list[3:])

# Lấy các phần tử trong khoảng từ index 2 đến trước index 5
print(number_list[2:5])

# Tạo một bản sao chép toàn bộ danh sách [:]
print(number_list[:])




```

```python
# Tìm vị trí index của phần tử dựa theo giá trị bằng index()

number_list = [5, 2, 8, 1, 9, 6, 4, 5, 9, 11, 12, 13]
print(number_list.index(5))
```

```python
# Cập nhật hoặc thêm nhiều cặp key-value vào từ điển bằng update()

dictionary = {"name": "Phong", "age": 30}
dictionary.update({"city": "Hà Nội", "postal_code": "01-001"})
print(dictionary)
```

```python
# Lấy danh sách tất cả các khóa (keys) trong từ điển

dictionary = {"name": "Phong", "age": 30, "city": "Hà Nội"}
keys = dictionary.keys()
print(keys)
```

```python
# Lấy danh sách tất cả các giá trị (values) trong từ điển

dictionary = {"name": "Phong", "age": 30, "city": "Hà Nội"}
dictionary2 = {"name": "Nam", "age": 31, "city": "Đà Nẵng"}

values = dictionary.values()
values2 = dictionary2.values()

print(values, values2)
```

```python
# Lấy tất cả các cặp khóa - giá trị bằng items()

dictionary = {
    "name": "Phong",
    "age": 30,
    "city": "Hà Nội",
    "name": "Nam",
    "age": 31,
    "city": "Đà Nẵng"
}

pairs = dictionary.items()
print(pairs)

```

```python
# Đóng gói các phần tử thành một tuple

tuple_example = 1, "apple", 1.12
print(tuple_example)

```

```python
# Hoán đổi giá trị giữa 2 biến cực kỳ nhanh trong Python

x = "apple"
y = "banana"

x, y = y, x
print(x, y)
```

```python
# Giải nén (unpack) các giá trị trong tuple ra từng biến riêng biệt

tuple_example = 1, "apple", 1.12
a, b, c = tuple_example

print(a)
print(b)
print(c)

```

```python
# Đếm số lần xuất hiện của một giá trị trong tuple bằng count()

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
count = tuple_test.count(2)
print(count)
```

```python
# Lấy phần tử trong tuple theo vị trí chỉ mục

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
print(tuple_test[2])

```

```python
# Cắt lát tuple theo dải chỉ mục

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
print(tuple_test[1:3])
```

```python
# Tìm vị trí của phần tử trong tuple bằng index()

tuple_test = (1, 2, 3, 4, 4, 3, 2, 2, 2)
tuple_index = tuple_test.index(4)
print(tuple_index)
```

```python
# Thêm phần tử vào tập hợp (set) bằng add() - set tự lọc trùng

set_example = {1, 2, 3}
set_example.add(4)
print(set_example)
```

```python
# Xóa phần tử khỏi tập hợp bằng remove()

set_example = {1, 2, 3}
set_example.remove(2)
print(set_example)
```

```python
# Phép hợp 2 tập hợp (union): lấy tất cả phần tử không trùng lặp

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = a.union(b)
print(c)
```

```python
# Phép giao 2 tập hợp (intersection): chỉ lấy các phần tử chung

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = a.intersection(b)
print(c)
```

```python
# Phép hiệu 2 tập hợp (difference): phần tử có ở tập 1 nhưng không có ở tập 2

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

c = b.difference(a)
print(c)
```

```python
# Kiểm tra xem tập này có phải là tập con của tập kia không (issubset)

a = {1, 2, 3}
b = {1, 2, 3, 4, 5, 6}

print(a.issubset(b))
```
