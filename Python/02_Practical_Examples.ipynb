Bài 1: Tạo các biến age (tuổi), height (chiều cao), và name (tên) that store respectively: age (integer), height (floating-point number), and name (string).
Print a sentence containing all these variables, e.g.:


```python
age = 30        # Số nguyên
height = 1.75  # Số thực
name = "Phong"  # Chuỗi ký tự

print(f"{name} is {age} tuổi và cao {height} m tall.")
```

Bài 2: Tạo hai biến số a và b. Tính và in ra tổng, hiệu, tích, thương của chúng.


```python
a=15
b=22
print("a+b=",a+b)
print("a-b=",a-b)
print("a*b=",a*b)
print(f"a/b= {a/b:.2f}")
```

Bài 3: Viết chương trình hỏi năm sinh của người dùng rồi tính xem hiện tại họ bao nhiêu tuổi.

Use the current year as a constant, e.g. current_year = 2026.

Print the result in the following format:


```python
birth_year = input("Nhập năm sinh của bạn: ")
current_year = 2026

birth_year = int(birth_year)
difference = current_year - birth_year

print(f"In 2026, you will be {difference} years old.")
```
