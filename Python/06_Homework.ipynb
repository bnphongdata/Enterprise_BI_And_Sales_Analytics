```python
Bài 1: Viết chương trình hỏi người dùng số điểm thi đạt được (thang điểm từ 0 đến 100).
Dựa trên số điểm nhập vào, tiến hành xếp loại kết quả:

- 0 đến 49 điểm: "Rất tiếc, bạn chưa đủ điều kiện vượt qua."
- 50 đến 69 điểm: "Chúc mừng bạn đã đạt cấp độ Cơ bản!"
- 70 đến 84 điểm: "Làm tốt lắm, bạn đạt cấp độ Trung cấp!"
- 85 đến 100 điểm: "Xuất sắc, bạn đạt cấp độ Nâng cao!"
```

```python
points = int(input("Nhập số điểm của bạn (0-100): "))

if 0 <= points <= 49:
    print("Rất tiếc, bạn chưa đủ điều kiện vượt qua.")
elif 50 <= points <= 69:
    print("Chúc mừng bạn đã đạt cấp độ Cơ bản!")
elif 70 <= points <= 84:
    print("Làm tốt lắm, bạn đạt cấp độ Trung cấp!")
elif 85 <= points <= 100:
    print("Xuất sắc, bạn đạt cấp độ Nâng cao!")
else:
    print("Lỗi: Vui lòng nhập số điểm hợp lệ từ 0 đến 100.")
```

```python
Bài 2: Viết chương trình yêu cầu nhập 2 số a và b, sau đó chia a cho b.
Sử dụng khối try / except để bắt lỗi chia cho số 0 (ZeroDivisionError) giúp chương trình không bị văng lỗi.
```

```python
try:
    a = float(input("Nhập số thứ nhất a: "))
    b = float(input("Nhập số thứ hai b: "))

    result = a / b
    print(f"Kết quả phép chia a / b là: {result}")

except ZeroDivisionError:
    print("Lỗi: Không thể chia cho số 0, vui lòng kiểm tra lại số thứ hai!")
```

```python
Bài 3: Cho một danh sách các số nguyên ngẫu nhiên:
[1, 3, 4, 5, 64, 78, 6, 55, 66, 75, 31, 22]

Dùng vòng lặp for kết hợp điều kiện if để lọc và in ra các số chẵn trong danh sách.
```

```python
liczby = [1, 3, 4, 5, 64, 78, 6, 55, 66, 75, 31, 22]

for liczba in liczby:
    if liczba % 2 == 0:
        print(liczba)
```

```python
Bài 4: Viết một hàm tên là double_number nhận vào một số và trả về giá trị gấp đôi của số đó.
Gọi thử hàm với vài số khác nhau để kiểm tra kết quả.
```

```python
def double_number(number):
    return number * 2


print(double_number(2))
print(double_number(5))
print(double_number(10))
print(double_number(-3))
```

```python
Bài 5: Viết một hàm kiểm tra số chẵn tên là is_even.
Hàm nhận vào một số và trả về True nếu là số chẵn, trả về False nếu là số lẻ.
```

```python
def is_even(number):
    return number % 2 == 0


print(is_even(2))   # True
print(is_even(7))   # False
print(is_even(0))   # True
print(is_even(-4))  # True
```
